## 2. Release Approval Gates

The TP approval process is built from three stages: MV validation, correlation run and analysis, and (from ES1) WTL pilot and analysis. The process supports a 7-day release cycle.

### 2.1 Gate Timeline
| MV Validation | Correlation Run | Correlation Analysis | WTL Pilot | WTL Analysis | TP Release |
|---|---|---|---|---|---|
| Prerequisite | Post TP integration | Post approver notification | Pilot factory runs | Approver analysis | HVM TP release |
| Day 0 | ED-6 days | ED-5 days | ED-3 days | ED-2 days | ED |

### 2.2 Yield Gate (Kappa)
| TP Robustness | Kappa Coefficient | Lower 95% |
|---|---:|---:|
| Post PRQ | 0.90 | 0.88 |
| Pre PRQ | 0.85 | 0.80 |

### 2.3 Parametric Gate
| Metric | Shift Threshold | Outlier Threshold |
|---|---:|---:|
| Vmin | 2 mV mean shift | 30 mV |
| SICC | Up to 0.2% | 10% |
| Cdyn | Up to 0.3% | 3% |

### 2.4 General Gate Rules
- Correlation comparison must be versus the current POR TP in factory.
- Any parametric, yield, or binsplit deviation must be justified.
- Shift: absolute group median or mean delta exceeds threshold.
- Outlier: unit absolute delta from group median exceeds threshold.
- Correlation analysis approval is gating for WTL run.
- Post ES1, WTL pilot approval is required before TP release.

## 3. Data Sources and Inputs

### 3.1 TP Comparison Inputs
- Release candidate TP and current POR TP in factory (for comparison).
- Relevant TP changes list and MV validation VIDs.
- Correlation lots list for analysis.
- Baseline TP selection for NVLH official runs: previous official TP letter (example: 03M00 vs 03L00).

### 3.2 Correlation Lot Requirements
- Correlation lots are units tested in both POR TP and release candidate TP.
- Target sample: 100 units per DLCP.
- CMOS skew coverage target: 5 corners, 60 units per corner (per request).
- Replace repetitive HW fallout units (for example B15, B99, B8, B98, B88, B10, B31) after consistent repetitive fails.
- Refresh units on process retargeting (for example XLTM to XLTMM), stepping changes (for example A to B), or major sort TP changes.

### 3.3 WTL Pilot Requirements
- From ES1 milestone and onward, TP release requires WTL pilot at ATM.
- Target sample: 1000 fresh units per DLCP.
- Include units from different fabs.

### 3.4 Parametric Data Scope for This Automation
- Vmin shifts by domain and frequency.
- SICC shifts by required SICC columns.
- Limiter content identity changes.
- Runtime temperature (DTS) shifts when relevant.
- Additional TP features that can influence binsplit, yield, or parametric behavior.

## 4. Second-Level Analysis Rules

Second-level analysis is mandatory when a top-level gate is failed or when an unexpected shift appears.

### 4.1 Root-Cause Drilldown Requirements
- Lot-level breakdown for affected population.
- Domain/frequency/core breakdown for Vmin and limiter-sensitive signals.
- SICC per-column and per-lot shift breakdown.
- DTS (thermal) shift review for affected corners.
- Correlation of Vmin shift with limiter pattern changes and thermal instability evidence.

### 4.2 Approval Template Expectations by Owner
1. Principle decision: approved, additional data required, or not approved.
2. Supporting data:
   - List of lots used for analysis.
   - Emanating risk statement.
   - Relevant results and root-cause evidence.

### 4.3 Parametric-Specific Supporting Data
- Parametric shift analysis for Vmin, SICC, and Cdyn per domain/frequency.
- Runtime temperature shift evidence when relevant.
- Limiter content changes between TPs that can affect Vmin or interpolation tools.

## 6. Implementation Notes

### 6.1 Automation Trigger and Notification Inputs
Before approval analysis run, automation should include in its report request:
1. Release candidate TP and POR TP in factory.
2. List of TP changes.
3. MV validation VIDs.
4. Correlation lots list.

### 6.2 Special Features Requiring Extra Validation
For high-risk features (content addition, CSE, ITD model release, TTR), include extra validation on top of basic flow.

#### Content addition release
- Validate new content using shmoo, MAV, MV run, and correlation vs previous POR TP.
- Add limiter pattern analysis and runtime temperature analysis.
- Post-ramp: include S2T validation and full impact study (500 units).
- Validate full-content TP with 1000-unit pilot per DLCP compared to PreSi expectations.

#### ITD model release
- Prerequisite offline and online validation.
- Integrate to POR only after prerequisite approval.
- Include correlation and WTL with required location code where applicable.
- Include binsplit view of aggressive vs relaxed ITD completion ratio (when relevant).

#### New flow matrix release
- Full VF curve comparison.
- Limiter module and pattern analysis for changed corners.
- Thermal stabilization analysis for changed corners.

#### Test time reduction (TTR) release
- Prerequisites: TP content freeze and limiter patterns allocated to SRH lists.
- Correlation lots: 100 units per DLCP.
- CMOS skew correlation across all skew corners.
- Validate limiter identity did not change.
- Full 1000-unit per DLCP WTL pilot.

### 6.3 Decision and Signoff Rule
All factory TP releases require approval from parametric, yield, and binsplit owners; post PRQ, binsplit and parametric ownership transfers to DMO.
